```python
import numpy as np
import networkx as nx
import scipy.special as sc
import matplotlib.pyplot as plt
import math 
import os
import networkx as nx
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.offline as pyo
from scipy.special import hyp2f1
```

```python
def f_H(x):
    return np.where((x<-3) | (x>3),0,np.sqrt(3)*abs(x)/np.pi/(3+x**2)*\
                    hyp2f1(1/3,2/3,1,x**2*(9-x**2)**2/(3+x**2)**3))
```

```python
def lambda_k(k):
    return np.sqrt(1 + 4 * np.cos(k[0]/2)**2 + 4 * np.cos(k[0]/2) * np.cos(np.sqrt(3)*k[1]/2)), \
    -1 * np.sqrt(1 + 4*np.cos(k[0]/2)**2 + 4 * np.cos(k[0]/2) * np.cos(np.sqrt(3)*k[1]/2))

k_x_values = np.linspace(0,4*np.pi,10**3)
k_y_values = np.linspace(0,4*np.pi/np.sqrt(3),10**3)

lambda_values = []
for j in range(len(k_x_values)):
    for jj in range(len(k_y_values)):
        temp1, temp2 = lambda_k([k_x_values[j], k_y_values[jj]])
        lambda_values.append(temp1)
        lambda_values.append(temp2)

x_values = np.linspace(-3,3,10**5)
y_values = f_H(x_values)

# Plot histogram
fig = px.histogram(lambda_values,
                   nbins=500,                  # Number of bins
                   histnorm='probability density'  # Normalize to probability density
                   )

fig.add_trace(go.Scatter(
    x=x_values,
    y=y_values,
    mode='lines',  # Use 'lines' or 'lines+markers' for different styles
    name='Function values',
    marker=dict(size=6, color='red')  # Customize marker size and color
))

# Update layout to fix title display and centering
fig.update_layout(
    xaxis_title = r"$\lambda(k_x,k_y)$",
    yaxis_title="Relative frequency",
    showlegend=False  # Deactivate legend
)

fig.update_xaxes(
        title_text = r"$\lambda(k_x,k_y)$",       # Modify the y-axis title text here
        title_font = dict(size=20),     # Set font size for y-axis title
        range = [-3.1,3.1],
        tickfont=dict(family="Arial Black", size=18)
    )
fig.update_yaxes(
        title_text=r"Relative frequency", # Modify the y-axis title text here
        title_font=dict(size=20),     # Set font size for y-axis title
        range=[0, 1],
        tickfont=dict(family="Arial Black", size=18)
    )

fig.write_image("Eigenvalue_frequency.pdf", 
                    format = "pdf", 
                    scale = 2, 
                    width=1000, 
                    height=500)

# Display the figure
#fig.show()
```
