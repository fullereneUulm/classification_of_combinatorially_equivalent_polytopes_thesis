```python
# Define the folder path of the .txt files
path_data = '/Users/arturbille/Documents/GitHub/classification_of_combinatorially_equivalent_polytopes_using_modern_statistical_methods_and_applications/character_of_fullerenes/character_results/alpha1:2_beta1:4'
```

```python
# Import the character data and centralize and normalize it

character_T_60 = np.loadtxt(path_data + '/character_T_60.txt')
character_T6_60 = np.loadtxt(path_data + '/character_T6_60.txt')

character_T_80 = np.loadtxt(path_data + '/character_T_80.txt')
character_T6_80 = np.loadtxt(path_data + '/character_T6_80.txt')

character_T_140 = np.loadtxt(path_data + '/character_T_140.txt')
character_T6_140 = np.loadtxt(path_data + '/character_T6_140.txt')

# Normalize and centralize the data
character_T_60_norm_centr = (character_T_60 - np.min(character_T_60))/(np.max(character_T_60)-np.min(character_T_60))
character_T6_60_norm_centr = (character_T6_60 - np.min(character_T6_60))/(np.max(character_T6_60)-np.min(character_T6_60))

character_T_80_norm_centr = (character_T_80 - np.min(character_T_80))/(np.max(character_T_80)-np.min(character_T_80))
character_T6_80_norm_centr = (character_T6_80 - np.min(character_T6_80))/(np.max(character_T6_80)-np.min(character_T6_80))

character_T_140_norm_centr = (character_T_140 - np.min(character_T_140))/(np.max(character_T_140)-np.min(character_T_140))
character_T6_140_norm_centr = (character_T6_140 - np.min(character_T6_140))/(np.max(character_T6_140)-np.min(character_T6_140))
```

```python
# Graph T_n
# Create a Plotly figure
fig = make_subplots(rows=1, cols=3)

# Add histogram
fig.add_trace(go.Histogram(x = character_T_60_norm_centr, histnorm='probability density', nbinsx=1000, name='Norm. histo.', marker_color='blue', opacity=1), row=1, col=1)
fig.add_trace(go.Histogram(x = character_T_80_norm_centr, histnorm='probability density', nbinsx=1000, name='Norm. histo.', marker_color='red', opacity=1), row=1, col=2)
fig.add_trace(go.Histogram(x = character_T_140_norm_centr, histnorm='probability density', nbinsx=1000, name='Norm. histo.', marker_color='green', opacity=1), row=1, col=3)

# Add the pdf function
#fig.add_trace(go.Scatter(x=x_vals, y=pdf, mode='lines', name=r'$f_{\Lambda_{'+str(p)+r','+str(p)+r'}}$'))

# Customize the layout
fig.update_layout(
    title = r'$\Big T_{60}, T_{80}, T_{140}, $',
    template = 'plotly',
    #xaxis_title = r'$\text{Centralized and normalized character values of }C_{60}, C_{80}, C_{140}$',
    #yaxis_title = r'$\text{rel. frequency}$',
    showlegend = False,
    title_x = 0.5,  # Center the title
    width = 1200,   # Adjust the width as needed
    height = 600,
    #xaxis = dict(
    #    title=dict(
    #        x=0.1,  # Shift the title to the left
    #        xanchor='left'  # Anchor it to the left to avoid overlap
    #    )
        #tickvals = np.round(endpoints[1:],3),  # Specify x-tick positions
        #ticktext = xtick_labels,     # Specify corresponding LaTeX labels
        #tickmode = 'array',           # Ensure tickvals and ticktext are used as specified
        #tickangle = 90,  # Rotate x-axis tick labels by 90 degrees
        #range = [0,9]
        #tickfont=dict(
        #    family='Arial Black, sans-serif',
        #    size=14,  # Set the font size of x-axis tick labels
        #    color='black'
        #)
    #)
    #yaxis=dict(
    #    range=[0, 20],  # Restrict the y-axis range to [0, 1]
    #)
)


fig.write_image('character_figures/pdf_character_centr&norm_T_60_80_140.pdf')
# Show the plot
fig.show()
# Save the plot as a PDF file
```
