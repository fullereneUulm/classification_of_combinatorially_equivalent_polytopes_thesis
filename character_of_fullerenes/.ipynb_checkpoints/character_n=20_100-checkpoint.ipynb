```python
# Ran on a Windows computer
# Compute the (1,1/2)-character for all fullerenes in C_n with n = 20, 24, 26, ..., 100
# Each fullerene is represented by its dual graph Tn and its hexagonal subgraph Tn6.
# All results are saved separately for each n in a .txt file.
```

```python
import itertools
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd
import pickle
import os
from scipy.linalg import expm
```

```python
path_database = "/mnt/n/fullerene_data"
path_program = "/mnt/n/buckygen_planarread"
path_savedata = "N:/fullerene_data/characters"
```

```python
def character(A, alpha = 1, beta = 1/2):
    D = np.diag(np.sum(A,axis=0))
    return np.trace(expm(alpha*A + beta*D))

def adjacencyTn6(A):
    hexagon_indices = np.where(np.sum(A,axis=0)==6)[0]
    A6 = A[hexagon_indices,:]
    A6 = A6[:,hexagon_indices]
    return A6

def run_buckygen(n):
    ### Input: Feasible n, i.e., n is even and (22 <= n or n = 20)
    ### Runs buckygen using the terminal and creates a (unreadable) file "Cn_dual" in the folder "fullerene_data" given by path_database
    m = int(n / 2 + 2)
    os.system('wsl ~ -e sh -c "cd '+ path_program +'; ./buckygen '+ str(m) +' '+ path_database +'/C'+ str(n) +'_dual"')
    print("Buckygen done.")
    
def run_planarread(n):
    ### Input: n - feasible integer
    ### Apply planarread to create Python-readible file
    m = int(n / 2 + 2)
    os.system('wsl  ~ -e sh -c "cd '+ path_program +'; ./planarread_mod <'+ path_database +'/C'+ str(n) +'_dual >'+ path_database +'/C'+ str(n) +'_dual_read.txt"')
    os.system('wsl  ~ -e sh -c "cd '+ path_database +'; rm C'+ str(n) +'_dual')
    print("Planarread done.")

def readin_planarread(n):
    # np.loadtxt('N:/fullerene_data/C'+str(n)+'_dual_read.txt', delimiter = ' ', usecols=(0, 1, 2, 3, 4, 5))
    temp = np.genfromtxt('N:/fullerene_data/C'+str(n)+'_dual_read.txt', usecols=(0, 1, 2, 3, 4, 5), invalid_raise=False)
    os.system('wsl  ~ -e sh -c "cd '+ path_database +'; rm C'+ str(n) +'_dual_read.txt')
    print("Import done.")
    return temp
    
def planarread_to_A_tensor(planarread,m,iso):
    A_tensor = np.zeros((m,m,iso))
    for i in range(iso):
        A = planarread[i*m:(i+1)*m,:]
        for j in range(m):
            for col in range(6):
                if A[j,col] != 0:
                    A_tensor[j,int(A[j,col]-1),i] = 1
    return A_tensor
```

```python
n_vec = np.array(np.append([20],np.arange(24,101,2)))
for n in n_vec:
    #print(n)
    m = int(n/2+2)
    run_buckygen(n)
    run_planarread(n)
    planarread_cn = readin_planarread(n)
    iso = int(np.shape(planarread_cn)[0]/m)
    #print(iso)
    character_Tn = np.zeros(iso)
    character_Tn6 = np.zeros(iso)

    A_tensor = planarread_to_A_tensor(planarread_cn, m, iso) 

    for j in range(iso):
        A = A_tensor[:,:,j]
        A6 = adjacencyTn6(A)
        character_Tn[j] = character(A, alpha = 1, beta = 1/2)
        if n>20:
            character_Tn6[j] = character(A6, alpha = 1, beta = 1/2)

    np.savetxt(path_savedata + '/character_T_'+str(n)+'.txt', character_Tn, fmt='%f')
    if n>20:
        np.savetxt(path_savedata + '/character_T6_'+str(n)+'.txt', character_Tn6, fmt = '%f')
    print('Data saved for n = '+str(n)+'.')
    print('-------------------------------')
```
