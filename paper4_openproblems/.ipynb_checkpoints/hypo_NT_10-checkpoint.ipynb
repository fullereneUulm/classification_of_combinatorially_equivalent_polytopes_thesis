```python
# Packages
import numpy as np
from IPython.display import display, HTML
from scipy.optimize import root_scalar
import plotly
import plotly.graph_objs as go
import plotly.io as pio
import plotly.offline as pyo
from plotly.subplots import make_subplots
plotly.offline.init_notebook_mode()
display(HTML(
    '<script type="text/javascript" async src="https://cdnjs.cloudflare.com/ajax/libs/mathjax/2.7.1/MathJax.js?config=TeX-MML-AM_SVG"></script>'
))
```

```python
def Lambda_pq_star_sample(p,q,N):
    U = np.random.uniform(0,np.pi,N)
    J = np.random.randint(0, high = p+q, size = N)
    return 3+2*(np.cos(U) + np.cos((p*U+2*np.pi*J)/(p+q)) + np.cos((q*U-2*np.pi*J)/(p+q)))

def Arcsin_pdf(x_vals,a,b):
    x_vals = np.array([x_vals])
    cond1 = x_vals <b
    cond2 = a < x_vals
    return 1/np.pi/np.sqrt((x_vals[cond1 & cond2] - a) * (b - x_vals[cond1 & cond2]))
```

```python
p = 1
q = 0
N = 10**6

Lambda_10_sample = Lambda_pq_star_sample(p,q,N)
```

```python
fig = go.Figure()

# Add histogram
fig = go.Figure(data=[go.Histogram(x = Lambda_10_sample, histnorm='probability density', nbinsx=1000, name='Norm. histo.')])

a = np.min(Lambda_10_sample)
b = np.max(Lambda_10_sample)

x_vals = np.linspace(a,b,N)
pdf_vals = Arcsin_pdf(x_vals,a,b)

fig.add_trace(go.Scatter(x = x_vals, y = pdf_vals, mode='lines', name='Line Plot'))

tickvals = [a, b]
ticktext = [ r'$'+str(int(np.round(a)))+'$',
    r'$'+str(int(np.round(b)))+'$'
]

fig.update_layout(
    template = 'plotly',
    title = r'$\Lambda_{'+str(p)+str(0)+'}^*$',
    width = 1200,   # Adjust the width as needed
    height = 600,
    xaxis_title = r'$x$',
    yaxis_title = r'Probability',
    xaxis = dict(
        range = [a - 0.1, b + 0.1],
        tickvals=tickvals,
        ticktext=ticktext,
        tickmode='array'
    ),
    yaxis = dict(
        range = [0,1]
    )
)
fig.show()
# Save the plot as a PDF file
fig.write_image('N_'+str(p)+str(q)+'_pdf.pdf')
```

## $(p,q) = (1,1)$

```python
p = 1
q = 1
N = 10**6
U = np.random.rand(N)*np.pi

Lambda_11_sample = Lambda_pq_star_sample(p,q,N)
Lambda_11_sample1 = 4*np.cos(U)**2 + 4*np.cos(U) + 1
Lambda_11_sample2 = 4*np.cos(U)**2 - 4*np.cos(U) + 1
```

```python
def temp_func(x_vals):
    if not(isinstance(x_vals, np.ndarray)):
        x_vals = np.array([x_vals])

    result = np.zeros(len(x_vals))
    for jj, x in enumerate(x_vals):
        if 0 < x and x < 9:
            result[jj] += 1/np.pi/2/np.sqrt(3*x-x**2+2*x*np.sqrt(x))
        if 0 < x and x < 1:
            result[jj] += 1/np.pi/2/np.sqrt(3*x-x**2-2*x*np.sqrt(x))
    return result
```

```python
fig = go.Figure()

# Add histogram
fig = go.Figure(data=[go.Histogram(x = Lambda_11_sample1, histnorm='probability density', nbinsx=1000, name='Norm. histo.')])

a = np.min(Lambda_11_sample1)
b = np.max(Lambda_11_sample1)

x_vals = np.linspace(a,b,N)
pdf_vals = temp_func(x_vals)

#pdf_vals = 0.5 * Arcsin_pdf(x_vals,a,b)
#pdf_vals[0] += 0.5
    

fig.add_trace(go.Scatter(x = x_vals, y = pdf_vals, mode='lines', name='Line Plot'))

tickvals = [a,1, b]
ticktext = [ r'$'+str(int(np.round(a)))+'$', 1,
    r'$'+str(int(np.round(b)))+'$'
]

fig.update_layout(
    template = 'plotly',
    title = r'1. Case: $\Lambda_{'+str(p)+str(q)+r'}^*$ with $J=0$',
    width = 1200,   # Adjust the width as needed
    height = 600,
    xaxis_title = r'$x$',
    yaxis_title = r'Probability',
    xaxis = dict(
        range = [a - 0.1, b + 0.1],
        tickvals=tickvals,
        ticktext=ticktext,
        tickmode='array'
    ),
    yaxis = dict(
        range = [0,2]
    )
)
fig.show()
```

```python
def temp_func_2(x_vals):
    cond1 = x_vals < 1
    cond2 = x_vals > 0
    return 6/4/np.pi\
    /np.sqrt(3*x_vals[cond1 & cond2] - x_vals[cond1 & cond2]**2 - 2*x_vals[cond1 & cond2]*np.sqrt(x_vals[cond1 & cond2]))
```

```python
fig = go.Figure()

# Add histogram
fig = go.Figure(data=[go.Histogram(x = Lambda_11_sample1, histnorm='probability density', nbinsx=500, name='Norm. histo.')])

a = np.min(Lambda_11_sample1)
b = np.max(Lambda_11_sample1)

x_vals = np.linspace(a,b,N)
pdf_vals = temp_func_2(x_vals)

#pdf_vals = 0.5 * Arcsin_pdf(x_vals,a,b)
#pdf_vals[0] += 0.5
    

fig.add_trace(go.Scatter(x = x_vals, y = pdf_vals, mode='lines', name='Line Plot'))

tickvals = [a, b]
ticktext = [ r'$'+str(int(np.round(a)))+'$',
    r'$'+str(int(np.round(b)))+'$'
]

fig.update_layout(
    template = 'plotly',
    title = r'1. Case: $\Lambda_{'+str(p)+str(q)+r'}^*$ with $J=1$',
    width = 1200,   # Adjust the width as needed
    height = 600,
    xaxis_title = r'$x$',
    yaxis_title = r'Probability',
    xaxis = dict(
        range = [a - 0.1, b + 0.1],
        tickvals=tickvals,
        ticktext=ticktext,
        tickmode='array'
    ),
    yaxis = dict(
        range = [0,2]
    )
)
fig.show()
```
