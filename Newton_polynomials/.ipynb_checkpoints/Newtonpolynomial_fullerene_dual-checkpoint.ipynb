```python
import numpy as np
import uufullerene as uf
```

```python
def vertex_connectivity_to_adjacencymatrix(n,vertex_connectivity,dual_indikator):
    m = int(n/2+2)
    # Here, we consider only fullerene graphs (not their duals)
    if dual_indikator == False:
        A = np.zeros((n,n))
        for j in range(n):
            temp = vertex_connectivity[j]
            for jj in range(3):
                temp2 = int(temp[jj]-1)
                #print(j,temp2)
                A[j,temp2] = 1
                A[temp2,j] = 1
    # This case considers dual fullerene graphs
    else:
        A = np.zeros((m,m))
        for j in range(m):
            temp = vertex_connectivity[j]
            for jj in range(6):
                if temp[jj]!=0:
                    temp2 = int(temp[jj]-1)
                    A[j,temp2] = 1
                    A[temp2,j] = 1
    return A

```

Verification of
\begin{align*}
\mu_{2k}\left( A_n \right)
&= 
\mu_{k}\left(A_n^* + \frac{1}{2}D_n^*\right)\quad \text{and}\quad \mu_{2k+1}\left( A_n \right) = 0,\quad k\in\mathbb{N}_0,
\end{align*}
where $A_n$ is the adjacency matrix of a fullerene graph, and $A_n^*$ and $D_n^*$ denote the adjacency and degree matrix, respectively, of the corresponding dual fullerene graph. 

```python
# Import adjacency matrices of fullerene graphs in C_20, C_24, C_26 (each with one isomer)
n = 20
# C_20 fullerene graph
with open('C20_read.txt', 'r') as f:
    lines = f.readlines()

lines = lines[:-1]
Vertex_connectivity_C20 = np.array([list(map(float, line.split()[:3])) for line in lines])
A_C20 = vertex_connectivity_to_adjacencymatrix(n,Vertex_connectivity_C20, False)

# C_20 dual fullere graph
with open('C20_dual_read.txt', 'r') as f:
    lines = f.readlines()

lines = lines[:-1]
Vertex_connectivity_C20_dual = np.array([list(map(float, line.split()[:6])) for line in lines])

A_C20_dual = vertex_connectivity_to_adjacencymatrix(n,Vertex_connectivity_C20_dual,True)

####################################
n = 24
# C_24 fullerene graph
with open('C24_read.txt', 'r') as f:
    lines = f.readlines()

lines = lines[:-1]
Vertex_connectivity_C24 = np.array([list(map(float, line.split()[:3])) for line in lines])
A_C24 = vertex_connectivity_to_adjacencymatrix(n,Vertex_connectivity_C24, False)


# C_24 dual fullere graph
with open('C24_dual_read.txt', 'r') as f:
    lines = f.readlines()

lines = lines[:-1]
Vertex_connectivity_C24_dual = np.array([list(map(float, line.split()[:6])) for line in lines])

A_C24_dual = vertex_connectivity_to_adjacencymatrix(n,Vertex_connectivity_C24_dual,True)

####################################
n = 26
# C_26 fullerene graph
with open('C26_read.txt', 'r') as f:
    lines = f.readlines()

lines = lines[:-1]
Vertex_connectivity_C26 = np.array([list(map(float, line.split()[:3])) for line in lines])
A_C26 = vertex_connectivity_to_adjacencymatrix(n,Vertex_connectivity_C26, False)

# C_26 dual fullere graph
with open('C26_dual_read.txt', 'r') as f:
    lines = f.readlines()

lines = lines[:-1]
Vertex_connectivity_C26_dual = np.array([list(map(float, line.split()[:6])) for line in lines])

A_C26_dual = vertex_connectivity_to_adjacencymatrix(n,Vertex_connectivity_C26_dual,True)
```

```python
from IPython.display import display, Math

# Display the LaTeX formatted string
display(Math('C_{20}:'))
for k in range(13):
    display(Math('k = '+str(k)))
    display(Math('\mu_{'+str(k)+r'}(A_{20}) = '+str(np.trace(np.linalg.matrix_power(A_C20,k)))))
    display(Math('\mu_{'+str(k)+r'}(A_{20}^*+\frac{1}{2}D_{20}^*) = '+str(np.trace(np.linalg.matrix_power(A_C20_dual+0.5*np.diag(np.sum(A_C20_dual,axis=0)),k)))))
```
