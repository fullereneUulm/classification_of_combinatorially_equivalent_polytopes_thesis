```python
def randomCC(n,CC1,CC2,CC3,CC4,CC5,CC6):
    boo = True
    k_max = 0
    k = 0
    ### Determine which CCn we are testing
    if n == 1:
        comparewith = CC1
    elif n == 2:
        comparewith = CC2
    elif n == 3:
        comparewith = CC3
    elif n == 4:
        comparewith = CC4
    elif n == 5:
        comparewith = CC5
    elif n == 6:
        comparewiht = CC6

    ### Construct a random connected component
    while k <= k_max and boo:
        temp = np.zeros((n,n))
        done = False
        while not done:
            indexes = np.random.randint(low = 0, high = n, size = 2)
            if indexes[0] != indexes[1]:
                temp[indexes[0],indexes[1]] = 1
                temp[indexes[1],indexes[0]] = 1
            if nx.is_connected(nx.Graph(temp)):
                done = True
                #nx.draw(nx.Graph(temp))

                found = False
                i = 0
                while (not found) and (i < comparewith.shape[2]):
                    if nx.is_isomorphic(nx.Graph(temp),nx.Graph(comparewith[:,:,i])):
                        found = True
                    i += 1
                if not found:
                    print("New connected component found")
                    nx.draw(nx.Graph(temp))
        k += 1
    print("Done")

randomCC(4,CC1,CC2,CC3,CC4,CC5,CC6)
```
