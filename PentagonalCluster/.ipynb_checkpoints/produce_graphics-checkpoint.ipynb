```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import time
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px
```

```python
n_vec = np.append([20],np.arange(24,157,2))
df = pd.read_excel('result/PIP_results.xlsx', header = None)
iso = pd.read_csv("result/iso_list.csv", header = None, usecols = [1], nrows = len(n_vec)).to_numpy().reshape((len(n_vec)))
pt = pd.read_csv("result/partitions.csv", header = None).to_numpy()
```

```python
for i in range(77):
    PIP_wanted = pt[i,:]
    PIP_string = "("+str(int(PIP_wanted[0]))
    for j in range(1,12):
        PIP_string = PIP_string +", "+str(int(PIP_wanted[j]))
    PIP_string = PIP_string + ")"
    #PIP_wanted = np.zeros(12)
    #PIP_wanted[0] = 6
    #PIP_wanted[1] = 6
    data_abs = df.iloc[np.where(np.sum(pt == PIP_wanted,axis = 1)==12)[0][0],:].to_numpy(dtype = int).reshape((len(n_vec)))
    data_rel = data_abs/iso

    
    figure="some_figure.pdf"
    fig=px.scatter(x=[0, 1, 2, 3, 4], y=[0, 1, 4, 9, 16])
    fig.write_image(figure, format="pdf")
    time.sleep(2)

    
    # Create figure with secondary y-axis
    fig = make_subplots(specs=[[{"secondary_y": True}]])
    
    # Add traces
    fig.add_trace(
        go.Scatter(x=n_vec, y=data_abs, 
                   name = "Absolute frequency"), 
        secondary_y=False,
    )
    
    fig.add_trace(
        go.Scatter(x=n_vec, y=data_rel, 
                   name = "Relative frequency"),
        secondary_y=True,
    )
    
    # Add figure title
    fig.update_layout(legend=dict(
        yanchor="top",
        y=0.99,
        xanchor="left",
        x=0.01
    )
        #title_text = "Frequencies of PIP: "+PIP_string,
        #title_x = 0.5
    )
    
    # Set x-axis title
    fig.update_xaxes(title_text ="n")
    
    # Set y-axes titles
    fig.update_yaxes(title_text="Absolute frequency", secondary_y=False)
    fig.update_yaxes(title_text="Relative frequency", secondary_y=True)
    
    fig.write_image("images/freq_PIP"+str(i)+".pdf")
    #fig.show()
```

```python
i = 0
PIP_wanted = pt[i,:]
PIP_string = "("+str(int(PIP_wanted[0]))
for j in range(1,12):
    PIP_string = PIP_string +", "+str(int(PIP_wanted[j]))
PIP_string = PIP_string + ")"
#PIP_wanted = np.zeros(12)
#PIP_wanted[0] = 6
#PIP_wanted[1] = 6
data_abs = df.iloc[np.where(np.sum(pt == PIP_wanted,axis = 1)==12)[0][0],:].to_numpy(dtype = int).reshape((len(n_vec)))
data_rel = data_abs/iso

# Create figure with secondary y-axis
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Add traces
fig.add_trace(
    go.Scatter(x=n_vec, y=data_abs, 
               name = "Absolute frequency"), 
    secondary_y=False,
)

fig.add_trace(
    go.Scatter(x=n_vec, y=data_rel, 
               name = "Relative frequency"),
    secondary_y=True,
)

# Add figure title
fig.update_layout(legend=dict(
    yanchor="top",
    y=0.99,
    xanchor="left",
    x=0.01
),
    title_text = "Frequencies of PIP: "+PIP_string,
    title_x = 0.5
)

# Set x-axis title
fig.update_xaxes(title_text = "n")

# Set y-axes titles
fig.update_yaxes(title_text="Absolute frequency", secondary_y=False)
fig.update_yaxes(title_text="Relative frequency", secondary_y=True)
#fig.write_image("images/freq_PIP"+str(i)+".pdf")
fig.show()
```
