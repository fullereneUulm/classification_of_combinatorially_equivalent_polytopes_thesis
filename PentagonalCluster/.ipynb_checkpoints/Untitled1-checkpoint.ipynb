```python
import numpy as np
import ab_fullerene as ab
import time

def print_poly(coefs):
    for i in range(len(coefs) - 1):
        if coefs[i] != 0:
            print(str(int(coefs[i])) + " * w^{" + str(i), end="} + ")
    print(str(int(coefs[len(coefs)-1])) + " * w^{" + str(len(coefs)-1), end="} ")
```

It holds
\begin{align*}
\text{res }f(z_0) 
&= \sum_{j=0}^{2k} \sum_{k_1+\ldots+k_7=k} \binom{k}{k_1,\ldots,k_7}\binom{2(k_1+k_2)+k_3+k_4+k_6+5j}{6k} 3^{k_4} w^{2(k_1+k_3) + k_2 + k_4 + k_5 + 5j} \underbrace{\lim_{z\rightarrow\infty} z^{2(k_1+k_2)+k_3+k_4+k_6+5j-6k}}_{=\begin{cases}
1,\quad &\text{ if } 2(k_1+k_2)+k_3+k_4+k_6+5j-6k=0,\\
0,\quad &\text{ else}.
\end{cases}
}.
\end{align*}
From Mathematica, for different $k$ we got the following formulas for the residues
\begin{align*}
k=0:\quad \text{res }f(z_0) &= 1, \\
k=1:\quad \text{res }f(z_0) &= w^5 + 3 w^6 + w^7, \\
k=2:\quad \text{res }f(z_0) &= w^{10} + 8 w^{11} + 15 w^{12} + 8 w^{13} + w^{14},\\
k=3:\quad \text{res }f(z_0) &= w^{15}+15 w^{16} + 60 w^{17} + 93 w^{18} + 60 w^{19} + 15 w^{20} + w^{21}, \\
k=4:\quad \text{res }f(z_0) &= w^{20} + 24 w^{21} + 160 w^{22} + 456 w^{23} + 639 w^{24} + 456 w^{25} + 160 w^{26} + 24 w^{27} + w^{28}, \\
k=5:\quad \text{res }f(z_0) &= 2 w^{25} + 40 w^{26} +355 w^{27} + 1530 w^{28} + 3540 w^{29} + 4655 w^{30} + 3540 w^{31} + 1530 w^{32} + 355 w^{33} +40 w^{34} +2 w^{35},\\
\end{align*}

```python
def formula1(k):
    result = np.zeros(7*k+1)
    for j in range(2*k+1):
        for k1 in range(k+1):
            for k2 in range(k-k1+1):
                for k3 in range(k-k1-k2+1):
                    for k4 in range(k-k1-k2-k3+1):
                        for k5 in range(k-k1-k2-k3-k4+1):
                            for k6 in range(k-k1-k2-k3-k4-k5+1):
                                k7 = k - k1 - k2 - k3 - k4 - k5- k6
                                if 2*(k1+k2)+k3+k4+k6+5*j-6*k == 0:
                                    result[2*(k1 + k3) + k2 + k4 + k5 + 5*j] += int(ab.multinomial([k1,k2,k3,k4,k5,k6,k7])\
                                    *ab.binom(2*(k1 + k2) + k3 + k4 + k6 + 5*j, 6*k)\
                                    *3**(k4))
    return result

k = 5
ex_coefs = formula1(k)
print_poly(ex_coefs)
```

Check
\begin{align*}
\mu_k &= 
- \sum_{k_1 + \ldots + k_7 = k} \binom{k}{k_1,\ldots,k_7} 3^{k_4} \cdot 1\lbrace 5k-k_1-k_2+k_5+k_7 \equiv 0\text{ mod } 5,~ k_3+k_5 = k_2+k_6 \rbrace.
\end{align*}

```python
def lhs(k):
    result = 0
    for k1 in range(k+1):
        for k2 in range(k-k1+1):
            for k3 in range(k-k1-k2+1):
                for k4 in range(k-k1-k2-k3+1):
                    for k5 in range(k-k1-k2-k3-k4+1):
                        for k6 in range(k-k1-k2-k3-k4-k5+1):
                            k7 = k - k1 - k2 - k3 - k4 - k5 - k6
                            if np.mod(5*k-k1-k2+k5+k7,5)==0 and k3 + k5 == k2+k6:
                                print("k1 = "+str(k1))
                                print("k2 = "+str(k2))
                                print("k3 = "+str(k3))
                                print("k4 = "+str(k4))
                                print("k5 = "+str(k5))
                                print("k6 = "+str(k6))
                                print("k7 = "+str(k7))
                                print("----------------")
                                result += ab.multinomial([k1,k2,k3,k4,k5,k6,k7])*3**(k4)
    return result

k = 5
print(lhs(k))
```

```python

comp_time_new = np.zeros(15)
comp_time_old = np.zeros(15)
for k in range(15):

    start_time = time.time()
    lhs(k)
    end_time = time.time()
    comp_time_new[k] = end_time - start_time
    
    
    start_time = time.time()
    ab.moment_dual_nanotube(5,0,k)
    end_time = time.time()
    comp_time_old[k] = end_time - start_time 
```

```python
import matplotlib.pyplot as plt

x = np.arange(15)

plt.plot(x, comp_time_old, 'r', x, comp_time_new, 'b')
plt.savefig("comp_time.png")
plt.show()
```
